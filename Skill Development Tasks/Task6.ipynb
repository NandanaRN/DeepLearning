````python
#Step 1: Import Libraries
import torch
import numpy as np

torch.manual_seed(42)
````

````python
#Step 2: Create Sample Sequence
sequence = torch.randn(5,3)

print(sequence)
````

````python
#Step 3: Initialize Weights
input_size = 3
hidden_size = 4

W_i = torch.randn(input_size+hidden_size, hidden_size)
W_f = torch.randn(input_size+hidden_size, hidden_size)
W_o = torch.randn(input_size+hidden_size, hidden_size)
W_g = torch.randn(input_size+hidden_size, hidden_size)

b_i = torch.zeros(hidden_size)
b_f = torch.zeros(hidden_size)
b_o = torch.zeros(hidden_size)
b_g = torch.zeros(hidden_size)

````

````python
#Step 4: Activation Functions
def sigmoid(x):
    return 1 / (1 + torch.exp(-x))
def tanh(x):
    return torch.tanh(x)
````

````python
def tanh(x):
    return torch.tanh(x)
````

````python
#Step 5: LSTM Cell
def lstm_cell(x, h_prev, c_prev):

    combined = torch.cat((x, h_prev))

    i = sigmoid(combined @ W_i + b_i)

    f = sigmoid(combined @ W_f + b_f)

    o = sigmoid(combined @ W_o + b_o)

    g = tanh(combined @ W_g + b_g)

    c = f * c_prev + i * g

    h = o * tanh(c)

    return h, c
````

````python
#Step 6: Forward LSTM
def forward_lstm(sequence):

    h = torch.zeros(hidden_size)

    c = torch.zeros(hidden_size)

    outputs = []

    for x in sequence:

        h, c = lstm_cell(x, h, c)

        outputs.append(h)

    return torch.stack(outputs)
````

````python
forward_output = forward_lstm(sequence)

print(forward_output)
````

````python
#Step 7: Backward LSTM

# Process the sequence in reverse.

def backward_lstm(sequence):

    h = torch.zeros(hidden_size)

    c = torch.zeros(hidden_size)

    outputs = []

    for x in reversed(sequence):

        h, c = lstm_cell(x, h, c)

        outputs.append(h)

    outputs.reverse()

    return torch.stack(outputs)
````

````python
backward_output = backward_lstm(sequence)

print(backward_output)
````

````python
#Step 8: BiLSTM Output
bilstm_output = torch.cat(
    (forward_output, backward_output),
    dim=1
)

print(bilstm_output)
````
