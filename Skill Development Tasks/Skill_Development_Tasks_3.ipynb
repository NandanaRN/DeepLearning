# Task 3: Vectorized Convolutional Forward Pass via im2col Transformation

## Objective

Implement a 2D convolution layer from scratch using NumPy by transforming
sliding image windows into a matrix (im2col), allowing convolution to be
performed as a single matrix multiplication.

### Features
- Pure NumPy
- Multi-channel support
- Configurable stride
- Configurable padding
- Vectorized implementation (no nested convolution loops)

### Why im2col?

Instead of sliding a kernel across an image repeatedly, im2col converts all
image patches into columns, making convolution equivalent to one large matrix
multiplication. This is the optimization used internally by many deep learning
libraries.

```python
#Import Libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Create Sample Image
np.random.seed(42)

# One RGB image
image = np.random.rand(1, 3, 32, 32)

print("Image Shape:", image.shape)
```

```python
#Create Convolution Filters
out_channels = 8
in_channels = 3
kernel_size = 3

filters = np.random.randn(
    out_channels,
    in_channels,
    kernel_size,
    kernel_size
)

print("Filter Shape:", filters.shape)
```

```python
#Padding Function
def zero_pad(image, pad):

    return np.pad(
        image,
        (
            (0,0),
            (0,0),
            (pad,pad),
            (pad,pad)
        ),
        mode='constant'
    )
```

```python
#im2col Function
def im2col(images, kernel_size, stride=1, padding=0):

    images = zero_pad(images, padding)

    N, C, H, W = images.shape

    out_h = (H-kernel_size)//stride + 1
    out_w = (W-kernel_size)//stride + 1

    cols = []

    for y in range(out_h):

        for x in range(out_w):

            patch = images[
                :,
                :,
                y*stride:y*stride+kernel_size,
                x*stride:x*stride+kernel_size
            ]

            cols.append(
                patch.reshape(N,-1)
            )

    cols = np.stack(cols,axis=1)

    return cols
```

```python
#Convert Image into Columns
columns = im2col(
    image,
    kernel_size=3,
    stride=1,
    padding=1
)

print(columns.shape)
```

```python
#Flatten Filters
filter_matrix = filters.reshape(
    out_channels,
    -1
)

print(filter_matrix.shape)
```

```python
#Vectorized Convolution
output = np.matmul(
    columns,
    filter_matrix.T
)

print(output.shape)
```

```python
#Convert Back to Feature Maps
output = output.reshape(
    1,
    32,
    32,
    out_channels
)

output = np.transpose(
    output,
    (0,3,1,2)
)

print(output.shape)
```

```python
#Visualize Feature Maps
fig, axes = plt.subplots(2,4, figsize=(12,6))

for i in range(8):

    ax = axes[i//4, i%4]

    ax.imshow(
        output[0,i],
        cmap="viridis"
    )

    ax.set_title(f"Filter {i+1}")

    ax.axis("off")

plt.tight_layout()

plt.show()
```
