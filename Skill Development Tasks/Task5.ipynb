````python
#Step 1: Import Libraries
import torch
import numpy as np

torch.manual_seed(42)
````

````python
#Step 2: Create Sample Data
X = torch.randn(4, 3)

print(X)
````

````python
#Step 3: Batch Normalization (Forward)
def batch_norm_forward(X, gamma, beta, eps=1e-5):

    mean = X.mean(dim=0, keepdim=True)

    var = X.var(dim=0, unbiased=False, keepdim=True)

    X_norm = (X - mean) / torch.sqrt(var + eps)

    out = gamma * X_norm + beta

    return out, X_norm
````

````python
gamma = torch.ones(1, 3)

beta = torch.zeros(1, 3)
````

````python
output_bn, X_norm_bn = batch_norm_forward(X, gamma, beta)

print(output_bn)
````

````python
#Step 4: Layer Normalization (Forward)
def layer_norm_forward(X, gamma, beta, eps=1e-5):

    mean = X.mean(dim=1, keepdim=True)

    var = X.var(dim=1, unbiased=False, keepdim=True)

    X_norm = (X - mean) / torch.sqrt(var + eps)

    out = gamma * X_norm + beta

    return out, X_norm
````

````python
output_ln, X_norm_ln = layer_norm_forward(X, gamma, beta)

print(output_ln)
````

````python
#Step 5: Manual Backward Pass
dout = torch.ones_like(output_bn)
````

````python
def backward(dout, X_norm):

    dgamma = (dout * X_norm).sum(dim=0, keepdim=True)

    dbeta = dout.sum(dim=0, keepdim=True)

    return dgamma, dbeta
````

````python
dgamma, dbeta = backward(dout, X_norm_bn)

print("dgamma:")
print(dgamma)

print()

print("dbeta:")
print(dbeta)
````

````python
#Step 6: Test LayerNorm Backward
dgamma_ln, dbeta_ln = backward(dout, X_norm_ln)

print(dgamma_ln)

print(dbeta_ln)
````
